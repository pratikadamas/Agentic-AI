```python
import os
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from pathlib import Path
```

```python
# read pdf file
def read_pdf(file_path):
    loader = PyPDFLoader(file_path)
    documents = loader.load()
    print(f"Loaded {len(documents)} pages from {file_path}")
    return documents

# split text into chunks
def split_text(documents, chunk_size=1000, chunk_overlap=200):
    text_splitter = RecursiveCharacterTextSplitter(
        chunk_size=chunk_size,
        chunk_overlap=chunk_overlap,
        # 'separators' (plural) — or omit entirely since these are defaults
        separators=["\n\n", "\n", " ", ""],
        # Optional: use token-based length instead of character count
        # length_function=lambda x: len(x.split())  # word count
    )
    print(f"Splitting text into chunks of size {chunk_size} with overlap of {chunk_overlap}...")
    texts = text_splitter.split_documents(documents)
    return texts

```

### pdf load and chunk done

```python
documents = read_pdf("../pdfs/Toffoli.pdf")
texts = split_text(documents)
print(f"Created {len(texts)} text chunks.")
```
