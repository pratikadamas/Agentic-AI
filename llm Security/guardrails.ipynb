```python
import re
```

```python
def deterministic_guardrails(text:str)-> bool:
    """
    This function checks if the input text contains any potentially harmful or unsafe content.
    It returns True if the text is safe, and False if it contains any unsafe content.
    """
    # Define a list of unsafe patterns (this is just an example, you can expand it)
    unsafe_patterns = [
        r'\b(?:hack|exploit|malware|virus)\b',  # Unsafe keywords
        r'<script.*?>.*?</script>',              # HTML script tags
        r'(?i)drop\s+table',                     # SQL injection pattern
        r'(?i)union\s+select',                   # SQL injection pattern
        r'(?i)alert\s*\(.*?\)',                  # JavaScript alert
    ]
    
    text_input =[
        "How to hack into a system",
        "This is a safe message",
        "DROP TABLE users; --",
        "<script>alert('Hello');</script>",
        "Normal text without any issues",
        "What is the weather today?"
    ]
    print("Checking text inputs for unsafe content:")
    
    for pattern in unsafe_patterns:
        if re.search(pattern, text):
            return False  # Unsafe content found
        
    for input_text in text_input:
        for pattern in unsafe_patterns:
            if re.search(pattern, input_text):
                print(f"Unsafe content found in input: '{input_text}'")
                return False  # Unsafe content found in the list of inputs    
    
    return True  # No unsafe content found
```
