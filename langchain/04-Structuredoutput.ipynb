#### Structured output allows agents to return data in a specific, predictable format. Instead of parsing natural language responses, you get structured data in the form of JSON objects, Pydantic models, or dataclasses that your application can use directly.

### Pydentic

```python
import os
from langchain.chat_models import init_chat_model
Groq_api_key=os.getenv("GROQ_API_KEY")
google_apiPkey=os.getenv("GOOGLE_API_KEY")

model= init_chat_model("qwen/qwen3.6-27b", model_provider="groq", api_key=Groq_api_key)
print(model)
```

```python
from pydantic import BaseModel, Field

class Movie(BaseModel):
    """Information about a movie."""

    title: str = Field(description="Movie title")
    director: str = Field(description="Movie director")
    year: int = Field(description="Release year")
    rating: float = Field(description="IMDb rating out of 10")
```

```python
res=model.invoke("details about Raone moive")
print(res.content) ## more informations
```

```python
model_with_schema = model.with_structured_output(Movie)
res = model_with_schema.invoke(
    "Give title, director, year and rating of the movie Inception"
)

print(res)
```
